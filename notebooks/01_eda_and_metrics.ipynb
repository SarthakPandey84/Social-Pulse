# 01. Exploratory Data Analysis & Engagement Metrics

Explores social media performance metrics, distributions of views, likes, shares, and saves, and basic data health.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df_posts = pd.read_csv('../data/sample/sample_posts.csv')
print(f'Total posts: {len(df_posts)}')
df_posts.head()
```

```python
df_posts[['views', 'likes', 'comments_count', 'shares', 'saves', 'retention_rate']].describe()
```

```python
# Save to share ratio distribution
df_posts['save_to_share_ratio'] = df_posts['saves'] / df_posts['shares'].replace(0, 1)
df_posts.groupby('topic')['save_to_share_ratio'].mean().sort_values(ascending=False)
```
